```python
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path
from T_method import LayeredStructure
from My_plotter import Plotter, Style
from Global_optimizer import my_json_load 
from My_interactive_plotter import PlotlyStyle, PlotlyPlotter
import re
```

```python
MU_0 = 4e-7 * np.pi
EPS_0 = 8.8541878188e-12
ETA_0 = np.sqrt(MU_0/EPS_0)
C = 1/np.sqrt(MU_0*EPS_0)
```

```python
f_min = 3.3e9
f_max = 4.2e9
f_0 = (f_min + f_max)/2
print(f'центральная частота - {(f_0*1.e-9):.2f} ГГц')
lamb_0 = C/f_0
k_0 = 2*np.pi/lamb_0
print(f'длина волны на центральной частоте - {(lamb_0*1000):.2f} мм')
print(f'волновое число на центральной частоте - {(k_0):.2f} 1/м')
```

```python
folder = Path(r".\results\2_sheets_l_80")
st = PlotlyStyle()
st.width = 800
st.scrollZoom = False
st.reference_line_width = 3
st.ax_label_size = 24
st.legend_size = 20
pl = PlotlyPlotter(st)
df = np.linspace(-30, 30, 200)/100
for path in folder.iterdir():
    data = my_json_load(path)
    path_str = str(path)
    match = re.fullmatch(r'results\\2_sheets_l_(\d+)\\(.+)_seed_(\d+).json', path_str)
    length, mode, seed = match.group(1), match.group(2), match.group(3)
    alpha = data["best_alpha"]
    beta = data["best_beta"]
    alpha_l = data["best_alpha_l"]
    alpha_c = data["best_alpha_c"]
    dipole_shift = data["best_dipole_shift"]
    beta_d = data["best_beta_d"]
    structure = LayeredStructure(alpha, beta=beta, alpha_l=alpha_l, alpha_c=alpha_c, dipole_shift=dipole_shift, beta_d=beta_d)
    directivity = 10*np.log10(structure.directivity_two_sources_diagonal(df))
    pl.plot((df*f_0+f_0)*1e-9, directivity, label=f"{mode.upper()}")
pl.set_ylim((0, 25))
pl.hline(17.5, color='gray', linestyle="15px,6px", alpha=0.5)
pl.hline(19.5, color='gray', linestyle="15px,6px", alpha=0.5)
pl.vline(f_max*1.e-9, color='gray', linestyle="15px,6px", alpha=0.5)
pl.vline(f_min*1.e-9, color='gray', linestyle="15px,6px", alpha=0.5)
pl.vline((1)*f_0*1.e-9, color='orange', linestyle="15px,6px", alpha=0.5)
pl.set_title(f'Two layers width={length}mm, target directivity: 17.5–19.5 dBi')
pl.set_xlabel('Frequency (GHz)')
pl.set_ylabel('Directivity (dBi)')
pl.finalize()
pl.show()
```
