```python
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path
from T_method import LayeredStructure
from My_plotter import Plotter, Style
from Global_optimizer import my_json_load 
```

```python
MU_0 = 4e-7 * np.pi
EPS_0 = 8.8541878188e-12
ETA_0 = np.sqrt(MU_0/EPS_0)
C = 1/np.sqrt(MU_0*EPS_0)
print(MU_0)
print(EPS_0)
print(ETA_0)
print(C)
print(np.pi)
```

* Пиковое усиление: 18 дБи по крайней мере на одной частоте в диапазоне  3300-4200 МГц
* Полоса пропускания: не менее 3300~5000 МГц (определяемая по уровню коэффициентов отражения от портов антенны ниже -10 дБ)
* Снижение усиления в в полосе частот:  не более 2 дБи от требуемого значения 18 дБи для 3300-4200 МГц
* Поляризация антенны: двойная поляризация +-45° (уровень коэффициента эллиптичности подлежит определению)
* Изоляция портов: >15 дБ
* Габаритные размеры антенны: не более 90x150x40мм3
* Количество портов: не менее 4 (с двойной поляризацией и перестраиваемой конфигурацией)
* Полоса покрытия: в горизонтальной плоскости угловое покрытие 180° с направленностью не менее 5 дБи

```python
f_min = 3.3e9
f_max = 4.2e9
f_0 = (f_min + f_max)/2
print(f'центральная частота - {(f_0*1.e-9):.2f} ГГц')
a = 90e-3
b = 150e-3
c = 40e-3
lamb_0 = C/f_0
k_0 = 2*np.pi/lamb_0
print(f'длина волны на центральной частоте - {(lamb_0*1000):.2f} мм')
print(f'волновое число на центральной частоте - {(k_0):.2f} 1/м')
beta_sum_max = k_0 * c
print(f'максимальная сумма beta {beta_sum_max:.2f} рад')
print(f'требуемая полоса - {(f_max-f_min)/f_0*100:.2f} %')
print(f'размер структуры в длинах волн - {(a/lamb_0):.2f} x {(b/lamb_0):.2f} x {(c/lamb_0):.2f}')
print(f'размер структуры в электрических длинах - {(k_0*a):.2f} x {(k_0*b):.2f} x {(k_0*c):.2f}')
```

```python
def width(directivity, df, thr):
    l = 0
    while directivity[l] < thr:
        l += 1
    r = len(directivity)-1
    while directivity[r] < thr:
        r -= 1
    width = df[r] - df[l]
    return width, l, r
def beta_to_mm(beta):
    return beta/k_0*1000
```

```python
print(beta_to_mm(0.2), 'mm')
print(f"размер ячейки для достижения -40 дБ на расстоянии 0.2 рад {beta_to_mm(0.2)*np.log10(np.e)*2*np.pi/6} mm")
```

```python
st = Style()
fig, ax = plt.subplots()
pl1 = Plotter(ax, st)
seeds = [90]
df = np.linspace(-50, 50, 200)/100
for seed in seeds:
    path = Path(f"project/N=1_seed_{seed}.json")
    data = my_json_load(path)
    alpha = data["best_alpha"]
    beta = data["best_beta"]
    structure = LayeredStructure(alpha, beta=beta)
    directivity = 10*np.log10(structure.directivity(df))
    pl1.plot(df, directivity, label=f"16dBi, seed={seed}")
pl1.finalize()
pl1.set_ylim((0, 25))
print(f'alpha = {alpha}')
print(f'beta = {beta}')
print(f'сумма beta: {np.sum(np.array(beta))}')
width_cur, l, r = width(directivity, df, 16)
print(f'ширина на уровне 16 дБ: {width_cur*100:.2f} %')
ax.axhline(18, color='gray', linestyle='--', alpha=0.5)
ax.axhline(16, color='gray', linestyle='--', alpha=0.5)
ax.axvline(df[l], color='gray', linestyle='--', alpha=0.5)
ax.axvline(df[r], color='gray', linestyle='--', alpha=0.5)
plt.show()
```
