```python
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path
from T_method import LayeredStructure
from My_plotter import Plotter, Style
from Global_optimizer import my_json_load, OptimizationParameters
```

```python
seed = 90
path = Path(f"temp/de_16_seed_{seed}.json")
data = my_json_load(path)
params = np.array(data['best_alpha'] + data['best_beta'])
print(f'best beta = {params}')
level = data['best_score']
```

```python
optp = OptimizationParameters()
optp.n = 3
optp.bounds_beta = [(0.1, 3.0) for _ in range(3)]
optp.bounds_alpha = [(0.0, 20.0)]*3
bounds = optp.bounds
k = optp.k
b = optp.b
optp.sigma_thres = 15.4
optp.segment_width_percent = 30.0
func = optp.objective_function_single_segment
x = np.linspace(0, 1, 200)
norm_params = (params - b)/k
obj_func = np.zeros_like(x)
```

```python
i = 0
x_i = np.linspace(bounds[i][0], bounds[i][1], 200)
obj_func_i = np.zeros_like(x)
for j in range(len(x)):
    norm_params_i_j = np.copy(norm_params)
    norm_params_i_j[i] = x[j]
    obj_func_i[j] = func(norm_params_i_j)
fig, ax = plt.subplots()
ax.plot(x_i, -obj_func_i)
ax.set_title(f'alpha_{i+1}')
ax.set_ylabel('Objective_func')
ax.set_xlabel(f'alpha_{i+1}')
ax.axhline(level, linestyle='--', color = 'grey')
ax.axvline(params[i], linestyle='--', color = 'red')
#ax.set_ylim((-40, 20))
plt.show()
```
